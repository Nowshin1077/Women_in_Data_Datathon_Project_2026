### Import Libraries and Dataset

```python
import pandas as pd
import numpy as np
```

```python
# 1. Load Raw CoAD Dataset

df_coad_raw = pd.read_csv(r"D:\Plateful_of_Data_Datathon26\FAOSTAT_Dataset\FAOSTAT_CoAHD_Ghana.csv")

df_coad_raw
```

```python
df_coad_raw.info()
```

```python
# 2. Selecting Core Columns & Dropping Redundant Codes/Flags

cols_to_keep = ['Year', 'Item Code', 'Item', 'Unit', 'Value']
df_clean = df_coad_raw[cols_to_keep].copy()
df_clean.head()
```

### TABLE A: Annual Time-Series of Macro Affordability Metrics (2017–2025)

```python
# Extract Cost of a Healthy Diet in Local Currency (GHS)
cohd_ghs = df_clean[
    (df_clean['Item'] == 'Cost of a healthy diet (CoHD)') & 
    (df_clean['Unit'] == 'LCU/cap/d')
][['Year', 'Value']].rename(columns={'Value': 'CoHD_GHS_day'})
cohd_ghs
```

```python
# Extract Cost of a Healthy Diet in Purchasing Power Parity (Int$ PPP)
cohd_ppp = df_clean[
    (df_clean['Item'] == 'Cost of a healthy diet (CoHD)') & 
    (df_clean['Unit'] == 'Int$ (PPP) per person per day')
][['Year', 'Value']].rename(columns={'Value': 'CoHD_PPP_day'})

cohd_ppp
```

```python
# Extract Prevalence of Unaffordability (%)
pua = df_clean[
    df_clean['Item'] == 'Prevalence of unaffordability (PUA)'
][['Year', 'Value']].rename(columns={'Value': 'Prevalence_Unaffordable_Pct'})

pua
```

```python
# Extract Number of People Unable to Afford (Millions)
nua = df_clean[
    df_clean['Item'] == 'Number of people unable to afford a healthy diet (NUA)'
][['Year', 'Value']].rename(columns={'Value': 'People_Unable_Millions'})

nua
```

```python
# Merge into a clean time-series DataFrame
df_coad_ts = cohd_ghs.merge(cohd_ppp, on='Year') \
                     .merge(pua, on='Year') \
                     .merge(nua, on='Year')

df_coad_ts
```

```python
# Feature Engineering: Year-over-Year Growth & Total Inflation Multiplier
df_coad_ts['CoHD_GHS_YoY_Growth_%'] = np.round(df_coad_ts['CoHD_GHS_day'].pct_change() * 100, 2)
df_coad_ts['Cumulative_Inflation_Multiplier'] = np.round(df_coad_ts['CoHD_GHS_day'] / df_coad_ts['CoHD_GHS_day'].iloc[0], 2)

df_coad_ts
```
