# 1. Data Cleaning and Feature Extraction 

#### The raw FAOSTAT CoAHD file contains multiple measurement units (LCU/cap/d, Int$ (PPP) per person per day, %, and million No) stacked in a long format. We clean, pivot, and engineer longitudinal affordability metrics across all years (2017–2025) and extract the granular 2021 food group cost structure

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# Configure styling
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.sans-serif': 'DejaVu Sans', 'font.size': 11})
```

```python
# 1. Load Raw CoAHD Dataset
df_raw = pd.read_csv(r"D:\Plateful_of_Data_Datathon26\Women_in_Data_Datathon_Project_2026\dataset\FAOSTAT_CoAHD_Ghana.csv")
df_coad_raw.head()
```

```python
# 2. Extract Essential Columns
cols = ['Year', 'Item Code', 'Item', 'Unit', 'Value']
df_clean = df_raw[cols].copy()
```

```python
# Feature Engineering
df_coad_ts['CoHD_GHS_YoY_Growth_%'] = np.round(df_coad_ts['CoHD_GHS_day'].pct_change() * 100, 1)
df_coad_ts['Nominal_Multiplier'] = np.round(df_coad_ts['CoHD_GHS_day'] / df_coad_ts['CoHD_GHS_day'].iloc[0], 2)
```

# 2. EDA (with Tabular outputs)

### TABLE A: Macro Affordability Time Series (2017–2025)


```python
cohd_ghs = df_clean[(df_clean['Item'] == 'Cost of a healthy diet (CoHD)') & 
                    (df_clean['Unit'] == 'LCU/cap/d')][['Year', 'Value']].rename(columns={'Value': 'CoHD_GHS_day'})
cohd_ppp = df_clean[(df_clean['Item'] == 'Cost of a healthy diet (CoHD)') & 
                    (df_clean['Unit'] == 'Int$ (PPP) per person per day')][['Year', 'Value']].rename(columns={'Value': 'CoHD_PPP_day'})
pua = df_clean[df_clean['Item'] == 'Prevalence of unaffordability (PUA)'][['Year', 'Value']].rename(columns={'Value': 'PUA_pct'})
nua = df_clean[df_clean['Item'] == 'Number of people unable to afford a healthy diet (NUA)'][['Year', 'Value']].rename(columns={'Value': 'NUA_millions'})

# Merge into unified analytical table
df_ts = cohd_ghs.merge(cohd_ppp, on='Year').merge(pua, on='Year').merge(nua, on='Year')

# Feature Engineering: Inflation multiplier, YoY growth, and implied exchange parity
df_ts['Nominal_Multiplier'] = np.round(df_ts['CoHD_GHS_day'] / df_ts['CoHD_GHS_day'].iloc[0], 2)
df_ts['YoY_Growth_Pct'] = np.round(df_ts['CoHD_GHS_day'].pct_change() * 100, 1)
df_ts['Implied_PPP_Rate'] = np.round(df_ts['CoHD_GHS_day'] / df_ts['CoHD_PPP_day'], 2)
```

```python
display(df_ts)
```

#### Key Insights Across All Years:
- Local Currency Escalation (+310.1%): Daily healthy diet expenditure quadrupled from 6.05 GHS to 24.81 GHS.
- The Structural Inflection Point (2022): Between 2017 and 2021, real diet costs in Int$ PPP were completely flat (~$3.46–$3.54). In 2022, the economic shield broke: local currency cost surged by +33.1% in a single year (reaching 11.43 GHS) and real international purchasing cost jumped to $3.83 PPP, marking the onset of the macro-inflationary squeeze.
- Entrenched Unaffordability: Unaffordability stayed locked above 64.7%, accumulating an additional 2.9 million excluded people by 2025 (23.1 million total).

### TABLE B: Food Group Cost Shares (2021 Benchmark)


```python
df_fg = df_clean[
    (df_clean['Year'] == 2021) & 
    (df_clean['Item'].str.startswith('Cost of')) & 
    (df_clean['Item'] != 'Cost of a healthy diet (CoHD)')
].pivot(index='Item', columns='Unit', values='Value').reset_index()

df_fg.columns = ['Item_Raw', 'Cost_PPP_day', 'Cost_GHS_day']
df_fg['Food_Group'] = df_fg['Item_Raw'].str.replace('Cost of ', '').str.title()
total_cohd_2021 = df_ts.loc[df_ts['Year'] == 2021, 'CoHD_PPP_day'].values[0]
df_fg['Basket_Share_Pct'] = np.round((df_fg['Cost_PPP_day'] / total_cohd_2021) * 100, 1)
df_fg = df_fg.sort_values(by='Cost_PPP_day', ascending=False).reset_index(drop=True)
```

#### Key Insights for HerPlate Hypothesis:
- Validating Dairy's Convergent Pressure: At 41.1% of healthy diet costs, animal-source foods create a prohibitive barrier that compounds milk supply contraction (-24.4%) to drive intake down to 15.4%.
- Validating Pulses' Supply Bottleneck: Legumes, nuts, and seeds cost only $0.39–$0.43 PPP/day (11.1% share). Low pulse consumption (23.5%) is therefore not an affordability failure—it is an upstream domestic supply collapse (-39.4%).

### TABLE C: 2022 Projected Food Group Cost Inflation (MDD-W Baseline)

```python
# 2022 aggregate healthy diet grew to 11.43 GHS ($3.83 PPP) from 8.59 GHS ($3.50 PPP) in 2021
inflation_factor_2022_ppp = df_ts.loc[df_ts['Year'] == 2022, 'CoHD_PPP_day'].values[0] / total_cohd_2021
inflation_factor_2022_ghs = df_ts.loc[df_ts['Year'] == 2022, 'CoHD_GHS_day'].values[0] / df_ts.loc[df_ts['Year'] == 2021, 'CoHD_GHS_day'].values[0]

df_fg_2022 = df_fg.copy()
df_fg_2022['2022_Cost_PPP_day'] = np.round(df_fg_2022['Cost_PPP_day'] * inflation_factor_2022_ppp, 2)
df_fg_2022['2022_Cost_GHS_day'] = np.round(df_fg_2022['Cost_GHS_day'] * inflation_factor_2022_ghs, 2)
```

# 3. Data Visualizations 

### Figure 1: Healthy Diet Cost Inflation in Ghana (2017–2025)

```python
plt.figure(figsize=(9, 5))
plt.plot(df_ts['Year'], df_ts['CoHD_GHS_day'], marker='o', color='#b03a2e', linewidth=2.5, label='Cost of Healthy Diet (GHS/day)')
plt.title('Cost Escalation of a Healthy Diet in Ghana (2017–2025)', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Year', fontweight='bold')
plt.ylabel('Cost per Person per Day (GHS)', fontweight='bold')
plt.xticks(df_ts['Year'])

# Highlight 2022 MDD-W survey year
plt.axvline(2022, color='#2c3e50', linestyle='--', linewidth=1.5, alpha=0.7)
plt.text(2022.1, 11.8, '2022 Survey Baseline:\n11.43 GHS ($3.83 PPP)', color='#2c3e50', fontweight='bold', fontsize=9.5)

for _, row in df_ts.iterrows():
    plt.annotate(f"{row['CoHD_GHS_day']:.2f}", (row['Year'], row['CoHD_GHS_day']),
                 textcoords="offset points", xytext=(0, 7), ha='center', fontsize=9)

plt.tight_layout()
plt.savefig(os.path.join(output_folder, "01_cohd_cost_escalation_2017_2025.png"), dpi=300, bbox_inches='tight')
plt.show()
```
