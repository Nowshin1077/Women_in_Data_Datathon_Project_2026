# 1. Data Cleaning and Feature Extraction 

#### The raw FAOSTAT CoAHD file contains multiple measurement units (LCU/cap/d, Int$ (PPP) per person per day, %, and million No) stacked in a long format. We clean, pivot, and engineer longitudinal affordability metrics across all years (2017–2025) and extract the granular 2021 food group cost structure

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import HTML, display
```

```python
# 1. Initialize output directory for saved visual assets
FIGURES_DIR = "figures"
os.makedirs(FIGURES_DIR, exist_ok=True)
```

```python
# 2. Load Raw CoAHD Dataset
df_raw = pd.read_csv(r"D:\Plateful_of_Data_Datathon26\Women_in_Data_Datathon_Project_2026\dataset\FAOSTAT_CoAHD_Ghana.csv")

df_raw.head()
```

```python
# 3. Select Core Attributes & Drop Metadata Noise
df_core = df_raw[['Year', 'Item', 'Unit', 'Value']].dropna(subset=['Value']).copy()
```

```python
# TABLE 1: Longitudinal Macro Affordability Time-Series (2017–2025)
# -------------------------------------------------------------------------
cohd_ghs = df_core[
    (df_core['Item'] == 'Cost of a healthy diet (CoHD)') & 
    (df_core['Unit'] == 'LCU/cap/d')
][['Year', 'Value']].rename(columns={'Value': 'CoHD_GHS_day'})

cohd_ppp = df_core[
    (df_core['Item'] == 'Cost of a healthy diet (CoHD)') & 
    (df_core['Unit'] == 'Int$ (PPP) per person per day')
][['Year', 'Value']].rename(columns={'Value': 'CoHD_PPP_day'})

pua = df_core[
    df_core['Item'] == 'Prevalence of unaffordability (PUA)'
][['Year', 'Value']].rename(columns={'Value': 'PUA_pct'})

nua = df_core[
    df_core['Item'] == 'Number of people unable to afford a healthy diet (NUA)'
][['Year', 'Value']].rename(columns={'Value': 'NUA_millions'})

# Merge into unified analytical table
df_macro_ts = cohd_ghs.merge(cohd_ppp, on='Year') \
                      .merge(pua, on='Year') \
                      .merge(nua, on='Year') \
                      .sort_values('Year') \
                      .reset_index(drop=True)

# Feature Extraction: Inflation Multipliers and YoY Escalations
base_cost_ghs = df_macro_ts.loc[df_macro_ts['Year'] == 2017, 'CoHD_GHS_day'].iloc[0]
df_macro_ts['Nominal_Inflation_Multiplier'] = np.round(df_macro_ts['CoHD_GHS_day'] / base_cost_ghs, 2)
df_macro_ts['YoY_Cost_Growth_Pct'] = np.round(df_macro_ts['CoHD_GHS_day'].pct_change() * 100, 1)

# Feature Extraction: Estimated Total Population and Excluded vs. Included Cohorts
df_macro_ts['Implied_Total_Pop_M'] = np.round(df_macro_ts['NUA_millions'] / (df_macro_ts['PUA_pct'] / 100.0), 2)
df_macro_ts['Affordable_Pop_M'] = np.round(df_macro_ts['Implied_Total_Pop_M'] - df_macro_ts['NUA_millions'], 2)
```

```python
print("--- TABLE 1: Cleaned Longitudinal Diet Affordability (2017–2025) ---")
display(df_macro_ts)
```

```python
# TABLE 2: Food Group Cost Composition (Reference Benchmark)
# -------------------------------------------------------------------------
df_fg = df_core[
    (df_core['Item'].str.startswith('Cost of ')) & 
    (df_core['Item'] != 'Cost of a healthy diet (CoHD)')
].pivot(index=['Year', 'Item'], columns='Unit', values='Value').reset_index()

df_fg.columns = ['Year', 'Raw_Item', 'Cost_PPP_day', 'Cost_GHS_day']
df_fg['Food_Group'] = df_fg['Raw_Item'].str.replace('Cost of ', '').str.title()

# Reference total CoHD in 2021: $3.50 PPP / 8.59 GHS
total_2021_ppp = 3.50
total_2021_ghs = 8.59

df_fg['Cost_Share_PPP_Pct'] = np.round((df_fg['Cost_PPP_day'] / total_2021_ppp) * 100, 1)
df_fg['Cost_Share_GHS_Pct'] = np.round((df_fg['Cost_GHS_day'] / total_2021_ghs) * 100, 1)

# Project food group costs to the 2022 MDD-W survey anchor year
# Macro scalar: 11.43 / 8.59 = 1.3306 for GHS; 3.83 / 3.50 = 1.0943 for PPP
scalar_ghs_2022 = 11.43 / 8.59
scalar_ppp_2022 = 3.83 / 3.50

df_fg['Projected_2022_GHS_day'] = np.round(df_fg['Cost_GHS_day'] * scalar_ghs_2022, 2)
df_fg['Projected_2022_PPP_day'] = np.round(df_fg['Cost_PPP_day'] * scalar_ppp_2022, 2)
df_fg = df_fg.sort_values(by='Cost_PPP_day', ascending=False).reset_index(drop=True)
```

```python
print("\n--- TABLE 2: Itemized Food Group Cost Basket Shares & 2022 Projections ---")
display(df_fg[['Food_Group', 'Cost_PPP_day', 'Cost_Share_PPP_Pct', 'Projected_2022_PPP_day', 'Projected_2022_GHS_day']])
```

### Key Takeways:
- Separating Mixed Metrics: The raw dataset placed daily costs in Ghanaian Cedis, international dollar benchmarks ($PPP), percentage rates, and population counts in millions inside a single data column. Leaving them mixed makes analysis prone to error. We separated each into its own column.

- Creating Growth Indicators: We calculated how many times diet costs expanded compared to 2017 (Nominal_Inflation_Multiplier) and calculated year-over-year escalation rates (YoY_Cost_Growth_Pct).

- Calculating Basket Shares: For each dietary component (such as animal-source foods, vegetables, legumes), we calculated the exact percentage of the daily healthy diet budget it consumes (Cost_Share_PPP_Pct). This directly populates the cost dimension ($B_i$) in HerPlate's bottleneck logic.

- 2022 Survey Baseline Projection: The MDD-W intake survey occurred in 2022. Because FAO disaggregates itemized food group costs in 2021, we mapped these component shares forward using the 2021-to-2022 macro cost shift (8.59 GHS $\rightarrow$ 11.43 GHS / $3.50 $\rightarrow$ $3.83 PPP) to identify prices during the exact period women's diets were measured.

### Key Findings:
- All-Years Affordability Squeeze (2017–2025): The cost of eating a healthy diet in Ghanaian Cedis escalated from 6.05 GHS/day in 2017 to 24.81 GHS/day in 2025. This represents a +310.1% increase (4.10x). While the proportion of people unable to afford this diet remained between 64.7% and 67.2%, population growth expanded the excluded population from 20.2 million to 23.1 million people.
  
- The 2022 MDD-W Survey Anchor Year: In 2022, daily diet costs reached 11.43 GHS ($3.83 PPP), an acceleration of +33.1% in a single year. Approximately 64.7% of the nation (21.50 million people) were financially priced out of a healthy diet in the year women's dietary intake was recorded.

- Component Cost Breakdown: Not all foods are equally responsible for this cost burden. Animal-Source Foods (ASF) require $1.44 PPP/day, consuming 41.1% of the entire healthy diet basket. By comparison, starchy staples represent 19.7%, vegetables 13.1%, and legumes/nuts/seeds only 11.1% ($0.39 PPP). This proves that while dairy and eggs face high cost barriers, pulses are not held back by retail price.



# 3. Data Visualizations


```python
# Configure professional charting aesthetic
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['DejaVu Sans', 'Arial', 'Helvetica'],
    'axes.edgecolor': '#cccccc',
    'axes.linewidth': 0.8,
    'grid.color': '#ebebeb',
    'grid.linestyle': '--'
})
```

```python
# FIGURE 1: All Years Diet Cost Escalation (2017–2025)

fig, ax = plt.subplots(figsize=(10, 5.5))
ax.plot(df_macro_ts['Year'], df_macro_ts['CoHD_GHS_day'], marker='o', color='#b03a2e', 
        linewidth=2.8, markersize=8, label='Daily Cost (GHS/day)')

# Annotate points
for _, row in df_macro_ts.iterrows():
    ax.annotate(f"{row['CoHD_GHS_day']:.2f} GHS\n({row['Nominal_Inflation_Multiplier']:.1f}x)", 
                (row['Year'], row['CoHD_GHS_day']),
                textcoords="offset points", xytext=(0, 10), ha='center', fontsize=9, fontweight='bold')

# Reference marker for 2022
ax.axvline(2022, color='#2c3e50', linestyle=':', linewidth=1.8, label='2022 MDD-W Survey Anchor')

ax.set_title('Escalation in Daily Cost of a Healthy Diet in Ghana (2017–2025)', fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel('Year', fontsize=11, fontweight='bold')
ax.set_ylabel('Cost per Person per Day (Ghana Cedis, GHS)', fontsize=11, fontweight='bold')
ax.set_ylim(0, 30)
ax.set_xticks(df_macro_ts['Year'])
ax.legend(loc='upper left', frameon=True)
fig.tight_layout()
fig.savefig(os.path.join(FIGURES_DIR, "01_coahd_all_years_cost_escalation.png"), dpi=300)
plt.show()
```

#### Figure 1 (Cost Trajectory): Visualizes the long-term inflation curve, marking the transition from steady prices to sharp escalation in 2022 and 2023.

```python
# FIGURE 2: All Years Population Unaffordability Headcount (2017–2025)

fig, ax = plt.subplots(figsize=(10, 5.5))
bar_colors = ['#2980b9' if yr != 2022 else '#d35400' for yr in df_macro_ts['Year']]
bars = ax.bar(df_macro_ts['Year'], df_macro_ts['NUA_millions'], color=bar_colors, width=0.55, edgecolor='#1b4f72', linewidth=1)

for bar, (_, row) in zip(bars, df_macro_ts.iterrows()):
    h = bar.get_height()
    ax.annotate(f"{h:.1f}M\n({row['PUA_pct']:.1f}%)", 
                (bar.get_x() + bar.get_width() / 2., h),
                ha='center', va='bottom', fontsize=9, fontweight='bold', xytext=(0, 3), textcoords='offset points')

ax.set_title('Population Unable to Afford a Healthy Diet in Ghana (2017–2025)', fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel('Year', fontsize=11, fontweight='bold')
ax.set_ylabel('Population Excluded (Millions)', fontsize=11, fontweight='bold')
ax.set_ylim(15, 27)
ax.set_xticks(df_macro_ts['Year'])
fig.tight_layout()
fig.savefig(os.path.join(FIGURES_DIR, "02_coahd_all_years_unaffordable_population.png"), dpi=300)
plt.show()
```

#### Figure 2 (Population Headcount): Shows the growing scale of affected individuals, demonstrating that a stable percentage (64.7%) translates to millions more excluded people over time due to population growth. The 2022 bar is highlighted in copper to anchor it to the survey period.

```python
# FIGURE 3: Food Group Cost Composition (HerPlate Cost Shares)

fig, ax = plt.subplots(figsize=(10, 5.2))
palette = ['#b03a2e', '#7f8c8d', '#27ae60', '#2980b9', '#f39c12', '#8e44ad']
bars = ax.barh(df_fg['Food_Group'], df_fg['Cost_Share_PPP_Pct'], color=palette, edgecolor='#333333', linewidth=0.8, height=0.55)

for bar, (_, row) in zip(bars, df_fg.iterrows()):
    w = bar.get_width()
    ax.annotate(f"{w:.1f}% (${row['Cost_PPP_day']:.2f} PPP)", 
                (w + 0.8, bar.get_y() + bar.get_height() / 2.), 
                va='center', fontsize=9.5, fontweight='bold', color='#2c3e50')

ax.set_title('Healthy Diet Basket Cost Drivers by Food Group (Reference Baseline)', fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel('Share of Total Daily Healthy Diet Basket (%)', fontsize=11, fontweight='bold')
ax.set_ylabel('')
ax.set_xlim(0, 50)
ax.invert_yaxis()
fig.tight_layout()
fig.savefig(os.path.join(FIGURES_DIR, "03_coahd_food_group_cost_shares.png"), dpi=300)
plt.show()
```

#### Figure 3 (Food Group Cost Shares): Displays the structural imbalance in healthy diet costs, where animal-source foods represent 41.1% of spending. This provides empirical backing for HerPlate's classification of Dairy under Convergent Pressure

```python
# FIGURE 4: Focus 2022 — Macro Diet Affordability Split

fig, ax = plt.subplots(figsize=(8, 5))
pop_2022_unable = df_2022_focus['NUA_millions'].iloc[0]
pop_2022_afford = df_2022_focus['Affordable_Pop_M'].iloc[0]

wedges, texts, autotexts = ax.pie(
    [pop_2022_unable, pop_2022_afford],
    labels=[f"Unable to Afford\n({pop_2022_unable:.1f}M / 64.7%)", f"Can Afford\n({pop_2022_afford:.1f}M / 35.3%)"],
    autopct='%1.1f%%',
    startangle=140,
    colors=['#c0392b', '#27ae60'],
    explode=(0.06, 0),
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
for at in autotexts:
    at.set_color('white')
    at.set_fontweight('bold')
    at.set_fontsize(11)

ax.set_title('Ghana Population Diet Affordability Divide (2022 MDD-W Baseline)', fontsize=13, fontweight='bold', pad=14)
fig.tight_layout()
fig.savefig(os.path.join(FIGURES_DIR, "04_coahd_2022_focus_population_split.png"), dpi=300)
plt.show()
```

#### Figure 4 (2022 Population Divide): A clean proportion chart showing that nearly two-thirds of Ghana was priced out of a healthy diet in the year women's dietary diversity was evaluated.

```python
# FIGURE 5: Focus 2022 — Projected Daily Food Group Outlays

fig, ax = plt.subplots(figsize=(10, 5.2))
asf_highlight = ['#b03a2e' if 'Animal' in fg else '#34495e' for fg in df_fg['Food_Group']]
bars = ax.barh(df_fg['Food_Group'], df_fg['Projected_2022_GHS_day'], color=asf_highlight, height=0.55)

for bar, (_, row) in zip(bars, df_fg.iterrows()):
    w = bar.get_width()
    ax.annotate(f"{w:.2f} GHS (${row['Projected_2022_PPP_day']:.2f} PPP)", 
                (w + 0.1, bar.get_y() + bar.get_height() / 2.), 
                va='center', fontsize=9.5, fontweight='bold')

ax.set_title('Projected 2022 Daily Expenditure by Food Group (Total: 11.43 GHS / $3.83 PPP)', fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel('Daily Expenditure (GHS per person per day)', fontsize=11, fontweight='bold')
ax.set_ylabel('')
ax.set_xlim(0, 6.0)
ax.invert_yaxis()
fig.tight_layout()
fig.savefig(os.path.join(FIGURES_DIR, "05_coahd_2022_focus_group_expenditures.png"), dpi=300)
plt.show()
```

#### Figure 5 (2022 Food Group Outlays): Translates relative basket shares into daily cash outlays (GHS), showing that an individual would have spent 4.70 GHS per day on animal-source foods alone out of an 11.43 GHS total budget.
